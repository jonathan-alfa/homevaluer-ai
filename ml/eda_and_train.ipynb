# House Price Prediction & Explainable AI (SHAP)
### Dataset: Kaggle House Prices - Advanced Regression Techniques (Ames Housing)

Notebook ini mendokumentasikan eksplorasi data (EDA), preprocessing fitur, perbandingan pelatihan model (Gradient Boosting & XGBoost Regressor), evaluasi metrics ($R^2$, RMSE, MAE), serta analisis kontribusi fitur lokal & global menggunakan **SHAP (SHapley Additive exPlanations)**.

```python
import os
import json
import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import shap

# 1. Load Data
df = pd.read_csv('../data/raw/train.csv')
print(f"Dimensi data: {df.shape}")
df.head()
```

## 1. Exploratory Data Analysis (EDA)
Mari kita lihat ringkasan target harga (`SalePrice`) dan distribusinya.

```python
print(df['SalePrice'].describe())

plt.figure(figsize=(10, 5))
sns.histplot(df['SalePrice'], kde=True, color='teal')
plt.title('Distribusi Harga Rumah (SalePrice) di Ames, Iowa')
plt.xlabel('Harga Jual ($)')
plt.ylabel('Frekuensi')
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Hubungan Fitur Kunci dengan SalePrice
- `OverallQual` vs `SalePrice`
- `GrLivArea` (Luas Bangunan) vs `SalePrice`

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))

sns.boxplot(x='OverallQual', y='SalePrice', data=df, ax=ax1, palette='viridis')
ax1.set_title('Hubungan Overall Quality vs SalePrice')
ax1.set_xlabel('Overall Quality (1-10)')

sns.scatterplot(x='GrLivArea', y='SalePrice', data=df, ax=ax2, alpha=0.6, color='coral')
ax2.set_title('Hubungan Luas Bangunan (GrLivArea) vs SalePrice')
ax2.set_xlabel('Living Area (sq ft)')

plt.tight_layout()
plt.show()
```
